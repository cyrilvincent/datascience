```python
# Loader le fichier
# Filtrer les données sur Absenteeism time in hours == 0 et les autres
# Moyenne et std du workload
# Rechercher tous les employés > 50 ans & workload > moyenne
# Dans les employés de plus de 50 ans & workload > moyenne prenez les 10 premiers (en position) et les 5 derniers (en position)
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/absenteeism_at_work/Absenteeism_at_work.csv", delimiter=";")
df.describe()
```

```python
df.columns = [col.strip().replace(" ", "_").lower() for col in df.columns]
df.columns
```

```python
df.dtypes
```

```python
no_absent = df[df["absenteeism_time_in_hours"] == 0]
no_absent.head(10)
```

```python
absent = df[df["absenteeism_time_in_hours"] != 0]
absent.head(10)
```

```python
absent_mean_workload = absent["work_load_average/day"].mean()
absent_std_workload = absent["work_load_average/day"].std()
absent_mean_workload, absent_std_workload
```

```python
no_absent_mean_workload = no_absent["work_load_average/day"].mean()
no_absent_std_workload = no_absent["work_load_average/day"].std()
no_absent_mean_workload, no_absent_std_workload
```

```python
df_sup50 = df[(df["age"] >= 50) & (df["work_load_average/day"] > absent_mean_workload)]
# df_sup50 = df[(df["age"] >= 50)][df["work_load_average/day"] > absent_mean_workload]
df_sup50
```

```python
# Créer une colonne absence_category
# Pour 0h, 4h, 8, 24, au delà (max)
# Faire un iterrows et n'afficher que les absence_category == 0
```

```python
absent_max = df["absenteeism_time_in_hours"].max()
absent_max
```

```python
bins = [-1, 0, 4, 8, 24, absent_max]
labels = ['Aucune', 'Courte (1-4h)', 'Demi-journée (5-8h)', 'Longue (9-24h)', 'Très longue (>24h)']
category = pd.cut(df["absenteeism_time_in_hours"], bins=bins, labels=False)
df["absence_category"] = category
# df["absence_category_label"]  = pd.cut(df["absenteeism_time_in_hours"], bins=bins, labels=labels)
df["absence_category"] = df["absence_category"].astype(np.int32)
# df["absence_category_label"] = df["absence_category_label"].astype(np.int32)
df.head(10)
```
