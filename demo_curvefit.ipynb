```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as opt
```

```python
noise = 10
def f(x):
    delta = (np.random.rand(x.shape[0]) - 0.5) * noise
    return 2.5 * x * np.sin(0.7 * x) + 2 + delta
```

```python
x = np.arange(0,10,0.1)
y = f(x)
```

```python
def poly4(x, a, b, c, d, e):
    return a * x ** 4 + b * x ** 3 + c * x ** 2 + d * x + e
```

```python
def xsinx(x, a, b, c):
    return a * x * np.sin(b * x) + c
```

```python
weight, cov = opt.curve_fit(poly4, x, y)
weight, cov
```

```python
weight_xsins, cov_xsinx = opt.curve_fit(xsinx, x, y)
weight_xsins, cov_xsinx
```

```python
plt.plot(x, poly4(x, weight[0], weight[1], weight[2], weight[3], weight[4]), color="red")
plt.plot(x, xsinx(x, weight_xsins[0], weight_xsins[1], weight_xsins[2]), color="black")
plt.scatter(x, y)
```
