```python
surfaces = data["np_surfaces"]
loyers = data["np_loyers"]
print(loyers)
```

```python
slope, intercept, rvalue, pvalue, stderr = stats.linregress(surfaces, loyers)
slope, intercept, rvalue, pvalue, stderr
```

```python
def poly2(x, a, b, c):
    return a * x ** 2 + b * x + c
```

```python
weight, cov = opt.curve_fit(poly2, surfaces, loyers)
weight, cov
```

```python
x = np.arange(400)
y = poly2(x, weight[0], weight[1], weight[2])
plt.scatter(surfaces, loyers)
plt.plot(surfaces, slope * surfaces + intercept, color="red")
plt.plot(x, y, color="black")
plt
```
