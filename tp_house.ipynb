```python
import numpy as np
import matplotlib.pyplot as plt
db = np.load("data/house/house.npz")
print(db)
```

```python
loyers = db["np_loyers"]
surfaces = db["np_surfaces"]
print(loyers)
print(surfaces)
```

```python
# Afficher toutes les surfaces > 200 : []
# Afficher les loyers dont la surface > 200 : []
# Créer la variable loyer_m2
# Afficher les loyer_m2 min & max
```

```python
print(surfaces[surfaces > 200])
```

```python
print(loyers[surfaces > 200])
```

```python
loyers_m2 = loyers / surfaces
loyers_m2[:10]
```

```python
print(np.min(loyers_m2), loyers_m2.max())
```

```python
# Affichant un nuage de point x=surface, en y les loyers
# idem mais que pour les surfaces < 100
```

```python
plt.scatter(surfaces, loyers)
plt.title("Loyers à Paris")
plt.show()
```

```python
surfaces_inf200 = surfaces[surfaces < 200]
loyers_inf200 = loyers[surfaces < 200]
```

```python
plt.scatter(surfaces_inf200, loyers_inf200)
plt.title("Loyers < 200m² à Paris")
plt.show()
```

```python
# Calculer la moyenne des loyers et surfaces
# Idem pour le std
# Filter les surfaces < mean + 3 * std
# Reafficher le nuage de points des data filtrées
```

```python
loyers_mean = np.mean(loyers)
loyers_std = np.std(loyers)
print(loyers_mean, loyers_std)
```

```python
surfaces_mean = np.mean(surfaces)
surfaces_std = np.std(surfaces)
print(surfaces_mean, surfaces_std)
```

```python
loyers_max = loyers_mean + 3 * loyers_std
loyers_filtered = loyers_inf200[loyers_inf200 < loyers_max]
surfaces_filtered = surfaces_inf200[loyers_inf200 < loyers_max]
plt.scatter(surfaces_filtered, loyers_filtered)
plt.title("Loyers < 200m² & 99.9% des loyers à Paris")
plt.show()
```
