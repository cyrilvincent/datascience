```python
import pandas as pd
import sklearn.linear_model as lm
import matplotlib.pyplot as plt
import sklearn.preprocessing as pp
import sklearn.pipeline as pipe
import numpy as np
```

```python
df = pd.read_csv("data/house/house.csv")
df.head(10)
```

```python
df = df[df["surface"] < 180]
```

```python
x = df["surface"].values.reshape(-1, 1)
y = df["loyer"]
```

```python
# Instanciation du modèle
model = lm.LinearRegression()
```

```python
# Apprentissage
model.fit(x, y)
```

```python
xpred = np.arange(180).reshape(-1, 1)
ypred = model.predict(xpred)
```

```python
model2 = pipe.make_pipeline(pp.PolynomialFeatures(2), lm.Ridge())
model2.fit(x, y)
ypred2 = model2.predict(xpred)
```

```python
plt.scatter(x, y)
plt.plot(xpred, ypred, color="red")
plt.plot(xpred, ypred2, color="black")
plt.show()
```
