```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("data/heartdisease/data_with_nan.csv", na_values=".")
# df.dtypes
df
```

```python
df.isnull().sum()
```

```python
# Pour slope, ca & thal droper les colonnes
# Pour chol créer une fonction qui créer mean + std * (rnd - 0.5) * 2 et l'appliquer avec la fonction apply
# .apply(lambda c: chol_mean + chol_std * (np.random.rand() - 0.5) * 2 if pd.isnull(c) else c)
# Pour trestbps, restecg, thalach & exang & fbs droper les lignes avec dropna
```

```python
df = df.drop(["slope", "ca", "thal"], axis=1)
df
```

```python
chol_mean = df["chol"].mean()
chol_std = df["chol"].std()
df["chol"] = df["chol"].apply(lambda c: chol_mean + chol_std * (np.random.rand() - 0.5) * 2 if pd.isnull(c) else c)
df
```

```python
df = df.dropna()
df
```

```python
df.to_csv("data/heartdisease/dataclean.csv", index=None)
```
