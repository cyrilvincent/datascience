```python
import pandas as pd
```

```python
depts = pd.read_csv("data/communes/dept.csv")
depts
```

```python
communes = pd.read_csv("data/communes/communes.csv", low_memory=False)
communes
```

```python
communes_depts = pd.merge(depts, communes, left_on="id", right_on="dept_id", suffixes=("", "_dept"))
communes_depts
```

```python
cps = pd.read_csv("data/communes/cp.csv")
cps
```

```python
communes_depts_cps = pd.merge(communes_depts, cps, left_on="code", right_on="code_insee", suffixes=("", "_cp"))
communes_depts_cps
```
