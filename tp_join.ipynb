```python
import pandas as pd
import numpy as np
```

```python
depts = pd.read_csv("data/communes/dept.csv", index_col="id")
depts.head(10)
```

```python
communes = pd.read_csv("data/communes/communes.csv", low_memory=False, index_col="id")
communes.head(10)
```

```python
depts["dept_id"] = depts.index
communes_depts = communes.join(depts, on="dept_id", rsuffix="_dept")
communes_depts.head(10)
```

```python
iris = pd.read_csv("data/communes/iris.csv", low_memory=False, index_col="id")
communes_depts_iris = pd.merge(communes_depts, iris, left_on="id", right_on="commune_id", suffixes=("", "_iris"))
communes_depts_iris.head(10)
```

```python
cps = pd.read_csv("data/communes/cp.csv", low_memory=False)
cps.head(10)
```

```python
final = pd.merge(communes_depts_iris, cps, left_on="code", right_on="code_insee", suffixes=("", "_iris"))
final
```

```python
cp = 38250
result = final[final["code_postal"] == cp]
result
```

```python
result.to_csv("data/communes/result.csv", index=False)
```
